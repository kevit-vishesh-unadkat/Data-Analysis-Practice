```python
import numpy as np

np.random.seed(42)

days = 30

daily_sales = np.random.randint(10_000, 50_001, size=days)

profit_margin = np.random.uniform(0.10, 0.25, size=days)

profit = daily_sales * profit_margin

print("Sales:")
print(daily_sales,end='\n\n')

print("Profit Margin:")
print(profit_margin,end='\n\n')

print("Profit:")
print(profit,end='\n\n')
```

```python
import pandas as pd

dates = pd.date_range(
    start="2024-01-01",
    end="2024-01-30"
)

regions = np.array([ "North", "South", "East", "West" ])

region_data = np.array([ regions[ i % 4 ] for i in range(30) ])

df = pd.DataFrame({
    "Date": dates,
    "Region": region_data,
    "Sales": daily_sales,
    "Profit_Margin": profit_margin,
    "Profit": profit
})

print(df)
```
