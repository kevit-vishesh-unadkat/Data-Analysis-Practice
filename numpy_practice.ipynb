```python
np.random.seed(40)
temp_c = np.random.uniform(20, 42, size=31)
temp_f = temp_c * 9 / 5 + 32
category = np.where(temp_c >= 35, "Hot", np.where(temp_c >= 25, "Warm", "Cool"))
print("Celsius:", temp_c)
print("Fahrenheit:", temp_f)
print("Categories:", category)
print("Hottest day:", np.argmax(temp_c) + 1)
print("Coolest day:", np.argmin(temp_c) + 1)
print("Average temperature:", temp_c.mean())
print("Hot/Warm/Cool counts:", *(np.count_nonzero(category == c) for c in ["Hot", "Warm", "Cool"]))
print("Days above average:", np.where(temp_c > temp_c.mean())[0] + 1)
print("Temperature range:", np.ptp(temp_c))
```
