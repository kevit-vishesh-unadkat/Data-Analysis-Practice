Pandas Series

```python
import pandas as pd
import numpy as np
```

```python
num=[10,23,46,7,89]
pd.Series(num)
```

```python
pd.Series(num,dtype="float")
```

```python
# custom Index

marks=[67,57,89,100]
subjects=['maths','english','ss','science']

pd.Series(marks,index=subjects)
```

```python
marks=pd.Series(marks,index=subjects,name='vishesh ke marks')
```
