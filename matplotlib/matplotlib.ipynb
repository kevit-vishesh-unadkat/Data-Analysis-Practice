```python
df=pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.plot(df['batsman'],df['2015'],marker="o",linewidth=3,label='2015')
plt.xlabel("name of players")
plt.ylabel("score of 2015 year")
plt.title("batsman season recorrd")

plt.plot(df['batsman'],df['2016'],marker="x",linewidth=3,label=2016)
plt.plot(df['batsman'],df['2017'],marker='+',linewidth=3,label=2017)

plt.legend()

plt.grid()
```
