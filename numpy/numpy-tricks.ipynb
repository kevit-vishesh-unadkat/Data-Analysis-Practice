```python
np.sort(b)

# in 2 d matrix by default sorting based on row wise
```

```python
np.sort(b,axis=0)

# to give axis = 0 then sorting based on column wise
```

```python
np.sort(b,axis=1)
# to give axis =1 then sorting based on row wise
```

```python
np.where(a>50)

# np. where function return the indeses of the element witch satisfied the specific condition 
```

```python
np.where(a>50,0,a)

# replace all value is greater then 50 replce with 0
```

```python
np.where((a>50)&(a%2==0),0,a)

# multiple condition is also allowd when we are using the where
```

```python
print(a)
print(np.argmax(a))

# argmax return the maximum element index 
```

```python
print(a)
print(np.argmin(a))

# argmin return the minimum element index
```

```python
print(b)
print(np.argmin(b,axis=0))
print(np.argmin(b,axis=1))
# axis=0 will give me a every column minimum element index
```

```python
# cumsum()

print(a)
print(np.cumsum(a))
```

```python
print(b)
print(np.cumsum(b))
# tihs will flat the array and convert it into 1D
print(np.cumsum(b,axis=0))
# this will calculate the cumsum column wise
print(np.cumsum(b,axis=1))
# this will calculate cumsum row wise
```

```python
print(a)
print(np.percentile(a,50))
print(np.percentile(a,5))
print(np.percentile(a,25))
```

```python
print(a)
print(np.histogram(a,bins=[0,50,100]))
```
