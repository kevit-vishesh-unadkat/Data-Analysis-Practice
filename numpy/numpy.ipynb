Array Atributes

```python
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
```

```python
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)

# ndim--> it will tell dimention of array
```

```python
print(a1.shape)
print(a2.shape)
print(a3.shape)  

# (2,2,2)  -->first 2 tell how many 2-D array and next two 2 represent no of rows and columns
```

```python
print(a1.size)
print(a2.size)
print(a3.size)

# size -->represent multiplication of rows and columns. 
```

```python
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
# int32 --> 4 byte lay che
# ane int64 -->8 byte lay che
```

```python
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

Stacking

```python
# Horizontal Stacking
import numpy as np
a4=np.arange(12).reshape(3,4)
print(a4)
a5=np.arange(12,24).reshape(3,4)
print(a5)
np.hstack((a4,a5))
```

```python
a4=np.arange(12).reshape(3,4)
print(a4)
a5=np.arange(12,24).reshape(3,4)
print(a5)
np.vstack((a4,a5))
```

```python
# flaten and revel

a=np.arange(12).reshape(3,4)
print(a)
b=a.flatten()
print(b)
b[0]=99
print(b)
```

```python
a=np.arange(12).reshape(3,4)
print(a)
b=a.ravel()
print(b)
b[0]=99
print(b)
```
